# DNA and Protein Analysis with BioAgent
This notebook demonstrates how to use the BioAgent for analyzing DNA and protein sequences.

````python
# Import necessary libraries
import sys
import os
sys.path.append(os.path.abspath(os.path.join('..')))

from bio_agent.core import BioAgent, SequenceType
from Bio.Seq import Seq
import matplotlib.pyplot as plt
%matplotlib inline
````

## DNA Analysis

````python
# Example DNA sequence
dna_sequence = "ATGCGATAGCTAGCTAGCTAGCATCGATCGATCGATCGATCGATCGATCGA"

# Initialize the agent
agent = BioAgent()

# Analyze the DNA sequence
dna_result = agent.analyze_sequence(dna_sequence, SequenceType.DNA)

# Print results
print(f"Sequence: {dna_result.sequence}")
print(f"Length: {dna_result.length} nucleotides")
print(f"GC Content: {dna_result.gc_content:.2f}%")
print(f"Molecular Weight: {dna_result.molecular_weight:.2f} g/mol")
print(f"Transcription (RNA): {dna_result.transcription}")
print(f"Translation (Protein): {dna_result.translation}")

# Get reverse complement
rc = agent.get_reverse_complement(dna_sequence)
print(f"\nReverse Complement: {rc}")
````

## Protein Analysis

````python
# Example protein sequence (human insulin)
protein_sequence = "MALWMRLLPLLALLALWGPDPAAAFVNQHLCGSHLVEALYLVCGERGFFYTPKTRREAEDLQVGQVELGGGPGAGSLQPLALEGSLQKRGIVEQCCTSICSLYQLENYCN"

# Analyze the protein sequence
protein_result = agent.analyze_sequence(protein_sequence, SequenceType.PROTEIN)

# Print results
print(f"Sequence: {protein_result.sequence}")
print(f"Length: {protein_result.length} amino acids")
print(f"Molecular Weight: {protein_result.molecular_weight:.2f} g/mol")
print(f"Isoelectric Point (pI): {protein_result.isoelectric_point:.2f}")

# Plot amino acid composition
agent.plot_amino_acid_composition(protein_result.amino_acid_composition)
````

## Using the Command Line Interface
You can also use the BioAgent from the command line. Here's how:

```bash
# Analyze a DNA sequence
python -m bio_agent.cli ATGCGATAGCTAGCTAGCTAGCATCGATCGATCGATCGATCGATCGATCGA --type dna

# Get reverse complement of a DNA sequence
python -m bio_agent.cli ATGCGATAGCTAGCTAGCTAGCATCGATCGATCGATCGATCGATCGATCGA --type dna --reverse-complement

# Analyze a protein sequence
```
